```python
# CONNECT WITH LLAMA LLM ON GROQ WITH YOUR API KEY
# 2. Import the library
from langchain_groq import ChatGroq

# 3. Connect to the LLM Brain (Open AI GPT OSS 120b)
llm = ChatGroq(
    api_key="YOUR API KEY", #<-- PASTE YOUR API KEY HERE
    model="openai/gpt-oss-120b"
)
```

```python
from IPython.display import Markdown

# 4. Updated prompt focused on MDM and LangGraph for a Technical Architect
prompt = """
Explain me the practical and brief roadmap to learn practical Agentic AI in 6 months. Give me as title with short explanation, not in tabular format or as paragraphs. Give me within 1000 tokens.
"""
response = llm.invoke(prompt,max_tokens=1000)

# 5. Print the LLM's response
print("🤖 Agentic AI Roadmap")
display(Markdown(response.content))
```
