### India Population Census Data Visualization
Below, we will plot the population of India as recorded during various census years.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Official Census Data of India (1901 to 2011)
census_data = {
    'Census Year': [1901, 1911, 1921, 1931, 1941, 1951, 1961, 1971, 1981, 1991, 2001, 2011],
    'Population (Millions)': [238.4, 252.09, 251.32, 278.98, 318.66, 361.08, 439.23, 548.15, 683.32, 846.42, 1028.73, 1210.85]
}

df_population = pd.DataFrame(census_data)

# Create a rich visualization
plt.figure(figsize=(12, 6))
sns.set_theme(style="whitegrid")

# Line plot with markers
ax = sns.lineplot(
    x='Census Year',
    y='Population (Millions)',
    data=df_population,
    marker='o',
    linewidth=2.5,
    color='#1f77b4'
)

# Highlight points with annotations
for index, row in df_population.iterrows():
    ax.text(
        row['Census Year'],
        row['Population (Millions)'] + 30,
        f"{row['Population (Millions)']:.1f}M",
        color='black',
        ha="center",
        fontsize=9
    )

plt.title('Population Growth of India (Census 1901 - 2011)', fontsize=16, fontweight='bold', pad=20)
plt.xlabel('Census Year', fontsize=12)
plt.ylabel('Population (in Millions)', fontsize=12)
plt.xticks(df_population['Census Year'])
plt.ylim(0, 1400)
plt.tight_layout()
plt.show()
```

### NSE Nifty 50 Index: 10-Year Historical Chart
We will now use the `yfinance` library to download and visualize the performance of the Nifty 50 index (`^NSEI`) over the last 10 years.

```python
!pip install -q yfinance
```

```python
import yfinance as yf
import matplotlib.pyplot as plt
import seaborn as sns
from datetime import datetime, timedelta

# Calculate date range for the last 10 years
end_date = datetime.today().strftime('%Y-%m-%d')
start_date = (datetime.today() - timedelta(days=365 * 10)).strftime('%Y-%m-%d')

# Download Nifty 50 data
nifty_df = yf.download('^NSEI', start=start_date, end=end_date)

# Create a styled plot
plt.figure(figsize=(14, 7))
sns.set_theme(style="darkgrid")

# Plot the Closing Price
plt.plot(nifty_df.index, nifty_df['Close'], label='Nifty 50 Close', color='#2ca02c', linewidth=2)

# Customize chart labels and title
plt.title('NSE Nifty 50 Index - Last 10 Years Performance', fontsize=16, fontweight='bold', pad=15)
plt.xlabel('Year', fontsize=12)
plt.ylabel('Index Points', fontsize=12)
plt.legend(loc='upper left', fontsize=11)
plt.tight_layout()
plt.show()
```
